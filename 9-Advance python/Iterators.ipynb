```python
class countdown:
    def __init__(self,start):
        self.start.currnt=start

    def __iter__(self):
        return self
    def __next__(self):
        if self.currnt<=0:
            raise StopIteration
        else:
            self.currnt-=1
            return self.currnt
        
for number in countdown(5):
    print(number)
```

```python
class car:
    def __init__(self,window,doors,engine_type):
        self.window=window
        self.doors=doors
        self.engine=engine_type

    def drive(self):
        print("car is self driving")

class tata(car):
    def __init__(self,window,doors,engine_type,self_driving):
        super().__init__(window,doors,engine_type)
        self.self_driving=self_driving

tata=tata(4,5,"petrol",True)
tata.drive()
print(tata.self_driving)
```

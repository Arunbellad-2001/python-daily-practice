```python
## employees Data Visualization
import pandas as pd
```

```python
data=pd.read_csv("employees.csv")
data.head(5)
```

```python
data.describe()
```

```python
data.info()
```

```python
# Remove missing values
df_clean = data.dropna(subset=["Experience_Years", "Salary"])

# Create figure
plt.figure(figsize=(8, 5))

# Scatter plot
plt.scatter(
    df_clean["Experience_Years"],
    df_clean["Salary"],
    marker="o",
    alpha=0.7
)

# Title
plt.title("Experience vs Salary")

# Axis labels
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")

# Grid
plt.grid(True, linestyle="--", alpha=0.5)

# Display
plt.show()
```

```python
df_clean = data.dropna(
    subset=["Experience_Years", "Salary", "Department"]
)

# Create figure
plt.figure(figsize=(8, 5))

# Get unique departments
departments = df_clean["Department"].unique()

# Plot each department
for department in departments:

    data = df_clean[df_clean["Department"] == department]

    plt.scatter(
        data["Experience_Years"],
        data["Salary"],
        label=department,
        alpha=0.7
    )

# Add title and labels
plt.title("Experience vs Salary by Department")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")

# Add legend
plt.legend()

# Add grid
plt.grid(True, linestyle="--", alpha=0.5)

# Display
plt.show()
```

```python
import matplotlib.pyplot as plt

# Remove missing values
df_clean = data.dropna(subset=["Experience_Years", "Salary"])

# Sort data by experience
df_clean = df_clean.sort_values("Experience_Years")

# Create line chart
plt.figure(figsize=(8, 5))

plt.plot(
    df_clean["Experience_Years"],
    df_clean["Salary"],
    marker="o"
)

# Add title and labels
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")

# Add grid
plt.grid(True)

# Display chart
plt.show()
```

```python
import matplotlib.pyplot as plt

# Remove missing values
df_clean = data.dropna(subset=["Experience_Years", "Salary"])

# Sort data by experience
df_clean = df_clean.sort_values("Experience_Years")

# Create bar chart
plt.figure(figsize=(10, 5))

plt.bar(
    df_clean["Experience_Years"],
    df_clean["Salary"],
    width=0.3
)

# Add title and labels
plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")

# Add grid
plt.grid(axis="y")

# Display chart
plt.show()
```

Data visualization with mathplolib

```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[2,3,4,5,6]

plt.plot(x,y, )
plt.title("line char")
plt.xlabel('x axis')
plt.ylabel('y axis')
plt.show()

```

```python
x=[1,2,3,4,5]
y=[1,4,12,16,25]

plt.plot(x,y,color='red',linestyle='--',linewidth=3,marker='o',markersize=9,)
```

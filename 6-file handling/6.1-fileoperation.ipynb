#### File Operation- Read And Write Files

File handling is a crucial part of any programming language. Python provides built-in functaions and methods to read from and write to files, both text and binary. This lesson will cover the basics of file handling, including reading and writing text files and binary files.

```python
names=[
    {"a=name":"arun","age":24,"course":"MCA"}
]
df=names
df
```

```python
### Read a Whole File
with open('exmaple.txt','r') as file:
    content=file.read()
    print(content)
```

```python
## Read a file line by line
with open('exmaple.txt','r') as file:
    for line in file:
        print(line.strip())
```

```python
## Writing a file(Overwriting)
with open('exmaple.txt','w') as file:
    file.write('we can to better\n')
    file.write('this is new line')

```

```python
with open('exmaple.txt','a') as file:
    file.write('\nwe can write without over writing\n')
```

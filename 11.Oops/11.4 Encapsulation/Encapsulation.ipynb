```python
class employeer:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def display(self):
        print(self.name)

emp=employeer("Ubed",25)
emp.display()
```

```python
class person:
    def __init__(self,name,age,gender):
        self.__name=name ## used privite method using __
        self.__age=age
        self.gender=gender

def get_name(person):
        return person.__name

person1=person("arun",25,"male")
get_name(person1)

## in this method we cant acces name becouse it is private

    
```

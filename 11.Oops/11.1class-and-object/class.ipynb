```python
### Instance Variable and Methods
class dog:
    ## constructor
    def __init__(self,name,age,type,color):
        self.name=name
        self.age=age
        self.type=type
        self.color=color

## create objects
dog1=dog("buddy",3,"hybrid","black")
print(dog1)
print(dog1.name)
print(dog1.age)
print(dog1.type)
print("dog color :",dog1.color)
```

```python
## Define a class with instance methods
class dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def bark(self):
        print(f"{self.name} say woof")

dog1=dog("buddy",3)
dog1.bark()
dog2=dog("lucy",5)
dog2.bark()
```

```python

## Define a class for bank account
class bankAccount:
	def __init__(self,owner,balance=0):
		self.owner=owner
		self.balance=balance

	def deposite(self,amount):
		self.balance+=amount
		print(f"{amount} is deposited.new balance is {self.balance}")
	
	def withdraw(self,amount):
		if amount>self.balance:
			print(f"insufficient balance.your balance is{self.balance}")
		else:
			self.balance-=amount
			print(f"{amount} is withdrawn.new balance is {self.balance}")
	
	def get_balance(self):
		return self.balance

account=bankAccount("Arun",100000)
print(account)
```

```python
account.deposite(10000000)
```

```python
account.withdraw(200000000)
```

```python
print(account.get_balance())
```

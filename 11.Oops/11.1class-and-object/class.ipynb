```python
### Instance Variable and Methods
class dog:
    ## constructor
    def __init__(self,name,age,type,color):
        self.name=name
        self.age=age
        self.type=type
        self.color=color

## create objects
dog1=dog("buddy",3,"hybrid","black")
print(dog1)
print(dog1.name)
print(dog1.age)
print(dog1.type)
print("dog color :",dog1.color)
```

```python
dog2=dog("julie",5,"female","white")

print("name :",dog2.name)
print("age :",dog2.age)
print("Gender :",dog2.type)
print("color",dog2.color)
```

```python
## Define a class with instance methods
class dog:
    def __init__(self,name,age):
        self.name=name
        self.age=age

    def bark(self):
        print(f"{self.name} say woof")

dog1=dog("buddy",3)
dog1.bark()
dog2=dog("lucy",5)
dog2.bark()
```

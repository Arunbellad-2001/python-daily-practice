```python
### Instance Variable and Methods
class dog:
    ## constructor
    def __init__(self,name,age,type):
        self.name=name
        self.age=age
        self.type=type

## create objects
dog1=dog("buddy",3,"hybrid")
print(dog1)
print(dog1.name)
print(dog1.age)
print(dog1.type)
```

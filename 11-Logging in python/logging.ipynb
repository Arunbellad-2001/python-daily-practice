```python
import logging

logging.basicConfig(
    filename='app.log',
    filemood='w',
    level=logging.DEBUG,
    formate='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datafmt='%y-%m-%d h%:m%:s%'
)

logging.debug("this is debug message")
logging.info("this is information message")
logging.warning("this is an waring message")
logging.error("this an error messgage")
logging.critical("this an critical message")
```

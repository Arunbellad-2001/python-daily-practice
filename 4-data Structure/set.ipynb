```python
## mathematical oper

set1={1,2,3,4,5}
set2={3,4,5,6,7,8}

## union oper
union_set=set1.union(set2)
print(union_set)

## intersection oper
intersection=set1.intersection(set2)
print(intersection)

## difference oper
difference=set1.difference(set2)
print(difference)

print(set1.difference(set2))
print(set2.difference(set1))
```

```python
## set methods
set1={1,2,3,4,5,}
set2={3,4,5}

## is subset
print(set1.issubset(set2))
print(set2.issubset(set1))
```

```python
## convert list of words into set to get unique words
text="convert list of words into set to get unique words"
word=text.split()

unique_word=set(word)
print(unique_word)
print(len(unique_word))
```

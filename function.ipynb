```python
def even_number(num):
    if num%2==0:
        return True
    else:
        return False

even_number(4)
```

```python
def area_of_circle(a,b,c):
    return 3.14*a*b*c

area_of_circle(2,3,4)
```

```python
def area_of_traiangle(a,b,c):
    return 0.5*a*b*c

area_of_traiangle(2,3,4)
```

```python
def evem_or_add(num):
    if num%2==0:
        return "even"
    else:
        return "odd"

evem_or_add(4)
```

```python
number1=int(input("enter first number"))
number2=int(input("enter second number"))
operation=input("+,- , *, /")

if operation=="+":
   print(number1+number2)
elif operation=="-":
   print(number1-number2)
elif operation=="*":
   print(number1*number2)
elif operation=="/":
   if number2!=0:
       print(number1/number2)
   else:
       print("division by zero is not allowed")
else:
    print("invalid operation")
print
```

```python
age=int(input("enter your age"))

if age<=7:
    price="free"
elif age<=12:
    price="100rs"
elif age<=25:
    is_student=input("you are student?(yes/no)").lower()
    if is_student=="yes":
        price="200rs"
    else:
        price="500rs"
elif age<=64:
    is_disbality=input("you are physically disablity?(yes/no)").lower()
    if is_disbality=="yes":
        price="250rs"
    else:
        price="500rs"
else:
    price="1000rs"
print("Ticket price is :",price)
```

```python
year=int(input("enter year:"))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year,"leaf year")
        else:
            print(year,"not leaf year")
    else:
        print(year,"leaf year")
else:
    print(year,"not leaf year")
```

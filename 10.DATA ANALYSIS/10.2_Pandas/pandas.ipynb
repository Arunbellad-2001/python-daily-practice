```python
import pandas as pd
```

```python
data=[1,2,3,4,5]
series=pd.Series(data)
print("series\n",series)
print(type(data))
print(type(series))
```

```python
## creating series from dictionary

data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
series_dict
```

```python
data=[10,20,30]
index_name=['a','b','c']
pd.Series(data,index=index_name)
```

```python
## creating dataframe form a dictionary list

data={
    'name':['arun','varun','shivu'],
    'age':[25,26,27],
    'city':['bengaluru','belagavi','hubbli']
}

df=pd.DataFrame(data)
print(df)
df
```

```python
### creating dataframe from list of dictionares
data=[
    {'name':'arun','age':25,'city':'bengaluru'},
    {'name':'varun','age':26,'city':'belagavi',},
    {'name':'shivu','age':27,'city':'hubli',}
]

df= pd.DataFrame(data)
print(df)
df
```
